```python
# BRONZE LAYER — Raw ingestion
# Read exactly as it comes — no changes

df_bronze=spark.read\
            .option("header","true")\
            .option ("inferSchema","true") \
            .option ("multiLine","true")\
            .option("escape",'"')\
            .option("quote",'"')\
            .csv("Files/flipkart_com-ecommerce_sample.csv")

print(f"Total Products: {df_bronze.count()}")
print(f"\nTotal Income: {len(df_bronze.columns)}")
print("\nColumn Names: ")
for col in df_bronze.columns:
    print(f" - {col}")
```

```python
df_bronze.select (
    "product_name",
    "product_category_tree",
    "retail_price",
    "discounted_price",
    "brand"
).show(3, truncate=80)
```

```python
df_bronze.write\
        .format("delta")\
        .mode("overwrite")\
        .saveAsTable("bronze_flipkart_products")

print("Bronze layer complete")
print(f"Total records saved: {df_bronze.count()}")
```
